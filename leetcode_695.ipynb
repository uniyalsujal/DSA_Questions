



# Number Of Islands

## Question
---

You are given an m x n binary matrix grid. An island is a group of 1's (representing land) connected 4-directionally (horizontal or vertical.) You may assume all four edges of the grid are surrounded by water.

The area of an island is the number of cells with a value 1 in the island.

Return the maximum area of an island in grid. If there is no island, return 0.



## Constraints
* m == grid.length
* n == grid[i].length
* 1 <= m, n <= 50
* grid[i][j] is either 0 or 1.

*  an island is formed when adjacent 1 are connected {up , down , right and left}. {same pattern as of flood fill}.

from above discussion i device a plan to  solve this problem  --
* firstly, i find out 1 in the grid which is simple by using double for loop.
* now if find out 1 that means that i find a land in it and call the dfs function.
* in it it converts 1 to 0 and whenever it do so it increase currentIslandArea by 1.
* we have four neighbours to check -
  * up = (row -1)(col)
  * down = (row +1)(col)
  * left = (row)(col-1)
  * right = (row)(col+1)
* and we have ranges of rows and col as -
  * 0 <= rows < grid.length
  * 0 <= cols < image[0].length  {these are the base cases of the recursion}
* at the end we return our currentIslandArea
* below is the mind map for counting areas of individual land

                    current cell

                    /    |    |    \
                   ↓     ↓    ↓     ↓

                    their areas
                         ↓
                     add them
                         ↓
                  return total

## Code
```
class Solution {
    int dfs(int[][] ar, int row, int col) {
        if ((row < 0 || row > ar.length - 1) || (col < 0 || col > ar[0].length - 1) || ar[row][col] == 0) {
            return 0;
        }
        ar[row][col] = 0;
        int currentIslandArea = 1;
        currentIslandArea = currentIslandArea + dfs(ar, row - 1, col);
        currentIslandArea = currentIslandArea + dfs(ar, row + 1, col);
        currentIslandArea = currentIslandArea + dfs(ar, row, col + 1);
        currentIslandArea = currentIslandArea + dfs(ar, row, col - 1);
        return currentIslandArea;
    }

    public int maxAreaOfIsland(int[][] grid) {
        int maxarea = 0;
        for (int i = 0; i < grid.length; i++) {
            for (int j = 0; j < grid[0].length; j++) {
                int islandarea = 0;

                if (grid[i][j] == 1) {
                    islandarea = dfs(grid, i, j);
                }
                maxarea = Math.max(islandarea, maxarea);
            }

        }
        return maxarea;
    }
}
```
